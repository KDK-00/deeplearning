```python
# Colab 환경에 필요한 라이브러리 설치
!pip install transformers

# transformers와 torch 라이브러리 불러오기
from transformers import MBartForConditionalGeneration, MBart50TokenizerFast
import torch

# 모델과 토크나이저 불러오기
model_name = "facebook/mbart-large-50-many-to-many-mmt"
model = MBartForConditionalGeneration.from_pretrained(model_name)
tokenizer = MBart50TokenizerFast.from_pretrained(model_name)

# 번역할 소스 언어와 타겟 언어 설정
src_text = "안녕하세요, 오늘 기분이 어떠세요?"
src_lang = "ko_KR"
tgt_lang = "en_XX"

# 소스 언어 설정
tokenizer.src_lang = src_lang

# 텍스트 토크나이즈
encoded_input = tokenizer(src_text, return_tensors="pt")

# 디코딩 시 필요한 타겟 언어 ID 설정
forced_bos_token_id = tokenizer.lang_code_to_id[tgt_lang]

# 번역 수행
translated_tokens = model.generate(**encoded_input, forced_bos_token_id=forced_bos_token_id)
translated_text = tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)[0]

# 번역 결과 출력
print(f"Translated text: {translated_text}")
```
